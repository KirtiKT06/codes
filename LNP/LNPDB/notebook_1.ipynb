```python
import pandas as pd

df = pd.read_csv("LNPDB.csv")

subset = df[
    (df["Experiment_method"] == "luminescence_normalized") &
    (df["Cargo"] == "mRNA") &
    (df["Cargo_type"] == "FLuc") &
    (df["Model"] == "in_vitro")
].copy()

print(subset.shape)

subset = subset.drop_duplicates(
    subset=["IL_SMILES"]
)

print(subset.shape)

subset.to_csv(
    "lnpdb_curated.csv",
    index=False
)
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np

smiles = subset["IL_SMILES"].tolist()

fps = []

for smi in smiles:

    mol = Chem.MolFromSmiles(smi)

    if mol is None:
        continue

    fp = AllChem.GetMorganFingerprintAsBitVect(
        mol,
        radius=2,
        nBits=2048
    )

    fps.append(np.array(fp))

X = np.array(fps)

print(X.shape)
```

```python
from sklearn.decomposition import PCA

pca = PCA(
    n_components=50,
    random_state=42
)

X_pca = pca.fit_transform(X)

print(
    pca.explained_variance_ratio_.sum()
)
```

```python
import umap

reducer = umap.UMAP(
    n_neighbors=30,
    min_dist=0.1,
    metric="jaccard",
    random_state=42
)

embedding = reducer.fit_transform(X)

print(embedding.shape)
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np

def analyze_cluster(cluster_id, subset):

    cluster_df = subset[subset["cluster"] == cluster_id].copy()

    def calc_desc(smiles):

        mol = Chem.MolFromSmiles(smiles)

        return pd.Series({
            "MW": Descriptors.MolWt(mol),
            "LogP": Descriptors.MolLogP(mol),
            "TPSA": Descriptors.TPSA(mol),
            "HBA": Lipinski.NumHAcceptors(mol),
            "HBD": Lipinski.NumHDonors(mol),
            "RotBonds": Lipinski.NumRotatableBonds(mol),
            "HeavyAtoms": mol.GetNumHeavyAtoms(),
            "RingCount": Lipinski.RingCount(mol)
        })

    desc = cluster_df["IL_SMILES"].apply(calc_desc)

    cluster_desc = pd.concat(
        [cluster_df.reset_index(drop=True),
         desc.reset_index(drop=True)],
        axis=1
    )

    # remove duplicate columns if present
    cluster_desc = cluster_desc.loc[
        :, ~cluster_desc.columns.duplicated()
    ]

    results = {}

    for descriptor in [
        "MW",
        "LogP",
        "TPSA",
        "HBA",
        "RotBonds",
        "HeavyAtoms"
    ]:

        X = cluster_desc[[descriptor]]
        y = cluster_desc["Experiment_value"]

        model = LinearRegression()
        model.fit(X, y)

        results[f"{descriptor}_R2"] = model.score(X, y)

    summary = {
        "cluster": cluster_id,
        "n": len(cluster_desc),
        "mean": cluster_desc["Experiment_value"].mean(),
        "std": cluster_desc["Experiment_value"].std(),
    }

    summary.update(results)

    return pd.Series(summary)
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski
import pandas as pd

def add_descriptors(df):

    def calc_desc(smiles):

        mol = Chem.MolFromSmiles(smiles)

        return pd.Series({
            "MW": Descriptors.MolWt(mol),
            "LogP": Descriptors.MolLogP(mol),
            "TPSA": Descriptors.TPSA(mol),
            "HBA": Lipinski.NumHAcceptors(mol),
            "HBD": Lipinski.NumHDonors(mol),
            "RotBonds": Lipinski.NumRotatableBonds(mol),
            "HeavyAtoms": mol.GetNumHeavyAtoms(),
            "RingCount": Lipinski.RingCount(mol)
        })

    desc = df["IL_SMILES"].apply(calc_desc)

    out = pd.concat(
        [df.reset_index(drop=True),
         desc.reset_index(drop=True)],
        axis=1
    )

    out = out.loc[:, ~out.columns.duplicated()]

    return out
```
