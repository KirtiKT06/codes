```python
import numpy as np
import pandas as pd
import MDAnalysis as mda

from pathlib import Path
from sklearn.decomposition import PCA
from scipy.stats import spearmanr, pearsonr
import matplotlib.pyplot as plt
```

```python
# ROOT = Path("/data/openmm_out/1PGA_mutations/run_330K")
ROOT = Path("/data/openmm_out/1PGA_mutations/explicit_solvent")

systems = {}

for folder in ROOT.iterdir():

    if not folder.is_dir():
        continue

    pdbs = list(folder.glob("*final.pdb"))
    dcds = list(folder.glob("*trajectory.dcd"))

    if len(pdbs) == 0 or len(dcds) == 0:
        continue

    systems[folder.name] = {
        "pdb": str(pdbs[0]),
        "dcd": str(dcds[0])
    }

print("Systems found:", len(systems))
```

```python
wt_name = "1PGA_rep1"

wt = mda.Universe(
    systems[wt_name]["pdb"],
    systems[wt_name]["dcd"]
)

wt_ca = wt.select_atoms("name CA")
```

```python
from MDAnalysis.analysis import align

trajectory_features = {}
all_frames = []

for name, files in systems.items():

    print(name)

    u = mda.Universe(
        files["pdb"],
        files["dcd"]
    )

    align.AlignTraj(
        u,
        wt,
        select="name CA",
        in_memory=True
    ).run()

    ca = u.select_atoms("name CA")

    frames = []

    for ts in u.trajectory[::10]:
        frames.append(
            ca.positions.flatten()
        )

    frames = np.array(frames)

    trajectory_features[name] = frames

    all_frames.append(frames)
```

```python
X = np.vstack(all_frames)

print(X.shape)

pca = PCA(n_components=5)

X_pca = pca.fit_transform(X)

print(
    "Variance:",
    pca.explained_variance_ratio_
)
```

```python
latent = {}

for name, frames in trajectory_features.items():

    latent[name] = pca.transform(frames)
```

```python
from scipy.stats import wasserstein_distance
import numpy as np

def ensemble_wasserstein(
    wt_latent,
    mut_latent,
    n_dims=5
):
    """
    Average Wasserstein distance across PCA dimensions
    """

    dists = []

    for i in range(n_dims):

        d = wasserstein_distance(
            wt_latent[:, i],
            mut_latent[:, i]
        )

        dists.append(d)

    return np.mean(dists)
```

```python
wt_latent = latent[wt_name]

wasserstein_scores = {}

for name, z in latent.items():

    if name == wt_name:
        continue

    score = ensemble_wasserstein(
        wt_latent,
        z,
        n_dims=5
    )

    wasserstein_scores[name] = score
```

```python
import numpy as np
from itertools import combinations

def extract_contact_features(ca_coords):

    """
    ca_coords:
        (n_frames, n_residues, 3)

    returns:
        (n_frames, n_pairs)
    """

    n_frames = ca_coords.shape[0]
    n_res = ca_coords.shape[1]

    pairs = list(combinations(
        range(n_res),
        2
    ))

    features = np.empty(
        (n_frames, len(pairs)),
        dtype=np.float32
    )

    for k, (i, j) in enumerate(pairs):

        diff = (
            ca_coords[:, i, :]
            -
            ca_coords[:, j, :]
        )

        features[:, k] = np.linalg.norm(
            diff,
            axis=1
        )

    return features
```

```python
import numpy as np

def get_local_feature_indices(pos, window=2):

    residues = range(
        max(2, pos-window),
        min(55, pos+window) + 1
    )

    idx = []

    for r in residues:

        col = r - 2

        idx.extend([
            col,          # sin(phi)
            col + 54,     # cos(phi)
            col + 108,    # sin(psi)
            col + 162     # cos(psi)
        ])

    return np.array(idx)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr

def plot_correlation(
    x,
    y,
    xlabel,
    ylabel="ΔΔG",
    title=None
):
    pearson = pearsonr(x, y)
    spearman = spearmanr(x, y)

    plt.figure(figsize=(6,5))

    sns.regplot(
        x=x,
        y=y,
        scatter_kws={"s":60},
        line_kws={"linewidth":2}
    )

    plt.xlabel(xlabel)
    plt.ylabel(ylabel)

    if title is None:
        title = xlabel

    plt.title(
        f"{title}\n"
        f"Pearson = {pearson.statistic:.3f} "
        f"(p={pearson.pvalue:.3g})\n"
        f"Spearman = {spearman.statistic:.3f} "
        f"(p={spearman.pvalue:.3g})"
    )

    plt.tight_layout()
    plt.show()
```
